## Dataset Description

### Ethics Discussion

- **Transparency & Explainability**: advanced techniques like deep learning achieve stellar accuracy, but such “black-box” models can be hard for clinicians to trust. ### Outliers Detection

```python
#outlier detection
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
# Loop through each category and feature
for category, suffixes in categories.items():
    print(f"\n=== {category} ===")
    for suffix in suffixes:
        for beat in ['0_', '1_']:
            col = f"{beat}{suffix}"
            if col not in df.columns:
                continue

            series = df[col].dropna()

            # --- IQR Method ---
            Q1 = series.quantile(0.25)
            Q3 = series.quantile(0.75)
            IQR = Q3 - Q1
            lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
            mask_iqr = (series < lower) | (series > upper)
            iqr_outliers = series[mask_iqr]

            # --- Z-score Method ---
            z_scores = np.abs(stats.zscore(series))
            mask_z = z_scores > 3
            z_outliers = series[mask_z]

            # Print counts
            print(f"{col:15s} | IQR outliers: {len(iqr_outliers):3d} | Z>3 outliers: {len(z_outliers):3d}")

    #(Optional) if you also want to *visualize* each feature’s outliers:
    fig, axes = plt.subplots(1, len(suffixes), figsize=(4*len(suffixes),4), constrained_layout=True)
    fig.suptitle(f"{category} — outlier histograms")
    for ax, suffix in zip(axes, suffixes):
         for beat, color in zip(['0_', '1_'], ['C0','C1']):
             col = f"{beat}{suffix}"
             if col not in df.columns:
                continue
             ax.hist(df[col].dropna(), bins=30, alpha=0.3, label=f"{beat} all", color=color)
             # overlay IQR outliers
             Q1, Q3 = df[col].quantile([0.25,0.75])
             IQR = Q3-Q1
             o = df[col][(df[col]<Q1-1.5*IQR)|(df[col]>Q3+1.5*IQR)]
             ax.hist(o, bins=10, alpha=0.8, label=f"{beat} IQR outliers", histtype='step')
         ax.set_title(suffix)
         ax.legend()
    plt.show()
```

Scaling our data is really essential and most importnatnly is doing is after splitting the dataset to avoid data leakage (Brownlee, 2020).

```python
from sklearn.preprocessing import StandardScaler

# 1. Instantiate the scaler
scaler = StandardScaler()

# 2. Fit on the training data only
scaler.fit(X_train)

# 3. Transform both training and test sets
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

### Data Augementation

Why SMOTE? SMOTE (Synthetic Minority Over-sampling Technique) shines when you have a moderately imbalanced dataset with enough minority‐class examples to interpolate between, but it can backfire if the minority class is extremely small, the feature space is sparse or high‐dimensional, or the data are noisy or categorical without proper preprocessing (Brownlee, 2021).

```python
#imbalance only on the training set 
from imblearn.over_sampling import SMOTE 
# 2) Apply SMOTE only to the training set
smote = SMOTE(random_state=42)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

print("Before SMOTE:", sorted(pd.Series(y_train).value_counts().items()))
print("After SMOTE: ", sorted(pd.Series(y_train_res).value_counts().items()))
```

# 1. SVM
according to literature, theSupport Vector Machines (SVMs) have proven highly effective for arrhythmia detection on standard ECG benchmarks like the MIT-BIH database, delivering robust decision boundaries even with limited or noisy data. Foundational work by Osowski et al. (2004) demonstrated reliable heartbeat recognition using SVMs with Hermite and higher‐order statistical features (Lin, Yang and Chiang, 2004), and later studies have enhanced performance further through optimized feature selection, advanced signal transforms, and hybrid deep–SVM architectures (Vinutha and Thirunavukkarasu, 2024)

```python
# Model pipelines and hyperparameter grids
pipeline_svc = Pipeline([
    ('scaler', StandardScaler()),
    ('classifier', SVC(random_state=42))
])

param_distributions_svc = {
    'classifier__C': [0.1, 1, 10],
    'classifier__gamma': [0.01, 0.1, 1],
    'classifier__kernel': ['linear', 'rbf']
}

# RandomizedSearchCV for SVC
random_search_svc = RandomizedSearchCV(
    pipeline_svc, param_distributions_svc, n_iter=10, scoring='accuracy', cv=cv, random_state=42
)
random_search_svc.fit(X_train_res, y_train_res)
best_svc = random_search_svc.best_estimator_
svc_test_score = accuracy_score(y_test, best_svc.predict(X_test))
results_svc.append(svc_test_score)

print("Test set accuracies for SVC:", results_svc)
```

## New approach for treating the Arrhythmia case

The high bias in the the arrhythmia dataset is a critical issue and reflects the real world special cases where we don't always have enough data for certain cases. However, this should encourage us to explore new ways to approach the problem and find better ways than the exisiting ones. 
// you can rely on the results that you find previously for ML/DL

Here’s a basic explanation for new approach that I propose: we want to classify three heartbeat types—N, Q, and VEB—but since Q and VEB are so rare (over 90k N beats versus only 15 Q beats), we’ll treat Q/VEB as “anomalies.” Our first model will decide if a beat is normal (N) or anomalous (Q/VEB), and then a second model will take the anomalies and classify them as Q or VEB. This way, we avoid trying to learn all three classes at once on such imbalanced data, and also because of the fact that creating synthetic data couldn’t simulate real-world inputs so the results wouldn’t be very accurate.

Also, since heartbeats are recorded continuously, I mean it’s not logical that we record a beat today and another one tomorrow, right?, it makes sense to think of this as a time-series problem rather than independent rows. We’d build an anomaly-detection model that looks at each beat in the context of its neighbors. One extra challenge is that the dataset spans many patients, each with their own ID—so we’d treat each patient’s record as a separate time series, but use the same model across all of them.


**What's promising about the idea**
- Tackles extreme imbalance head-on by reframing Q/VEB as anomalies. A two-stage setup means our first model only needs to spot “weird” beats, so it doesn’t drown in 90 k normals—then a lightweight second model teases apart the rare classes.
- Leverages temporal component. Treating each patient’s beats as a time series (rather than isolated rows) can give better results for pre- and post-beat cues that static classifiers miss—those tiny shifts in RR intervals.

**Key limitations to watch for**
- Only 44 patients, 44 “series.” Deep time-series models often crave hundreds of distinct sequences. With so few patients, you risk overfitting to idiosyncratic rhythms (age, electrode placement, comorbidities).
- Train/test split is tricky. If we mix beats from the same person into both sets, the performance will look unrealistically good (the model “recognizes” the patient, not the pathology). so we’ll need a strictly leave-one-patient-out or similar cross-validation to get a true sense of generalization.
- Anomaly thresholds can be brittle. What counts as “anomalous” may drift across subjects or recording conditions. Setting a single cutoff risks either swamping your secondary classifier with false alarms or missing subtle Q-beats altogether (OpenAI, 2025).

- Pipeline complexity vs. payoff. Two models plus sequence modeling demands more data wrangling, tuning, and compute. For only a few dozen rare events, the gains over a well-tuned Random Forest or CNN might be marginal—especially if your anomaly detector struggles in noisy segments (OpenAI, 2025).
